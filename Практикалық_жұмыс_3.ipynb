```python
# 2 тапсырама
def hill_climbing(values, start):
    current = start

    while True:
        neighbors = []

        if current > 0:
            neighbors.append(current - 1)

        if current < len(values) - 1:
            neighbors.append(current + 1)

        best = current

        for n in neighbors:
            if values[n] > values[best]:
                best = n

        if best == current:
            break

        current = best

    return current


values = [1, 3, 5, 8, 6, 4, 2]

for start in [0, 2, 4, 6]:
    index = hill_climbing(values, start)
    print(start, "->", index, "->", values[index])
#Барлық бастапқы индекстерді тексергенде нәтиже бірдей болды. Барлығында алгоритм 3-индексте тоқтап, мәні 8 болды. Сондықтан бұл жағдайда қай индекстен бастағанымыз нәтижеге әсер етпеді.
```

```python
# 3 тапсырма

values = [1, 4, 7, 5, 3, 6, 9, 8]

for start in [0, 4]:
    index = hill_climbing(values, start)
    print(start, "->", index, "->", values[index])

    #start = 0 болғанда алгоритм 7 мәніне дейін барып тоқтады. start = 4 болғанда 9 мәніне келді. Бұл жерде 7 — local maximum, ал 9 — global maximum. Бірінші жағдайда 9-ға жетпейді, себебі алгоритм тек қазіргі тұрған жерінің көршілерін қарайды.
```

```python
# 4 тапсырма
import random

def random_restart(values, attempts):
    best_position = None

    for _ in range(attempts):
        start = random.randint(0, len(values) - 1)
        position = hill_climbing(values, start)

        if best_position is None or values[position] > values[best_position]:
            best_position = position

    return best_position
```

```python
values = [1, 4, 7, 5, 3, 6, 9, 8]

for i in range(3):
    result = random_restart(values, attempts=5)
    print("Эксперимент", i + 1, ":", values[result])


    #Random Restart алгоритмі әр жолы әртүрлі бастапқы күйден басталып көреді. Сондықтан алгоритм бір ғана бастапқы нүктеде тоқтап қалмайды. Бірнеше бастапқы нүктені тексеру арқылы global maximum-ды табу мүмкіндігі артады.
```

```python
# 5 тапсырма


import random
import math

def random_neighbor(current, n):
    neighbors = []

    if current > 0:
        neighbors.append(current - 1)

    if current < n - 1:
        neighbors.append(current + 1)

    return random.choice(neighbors)


def anneal(values, start, T=10):
    current = start

    while T > 0.1:
        next_state = random_neighbor(current, len(values))
        dE = values[next_state] - values[current]

        if dE > 0 or random.random() < math.exp(dE / T):
            current = next_state

        T *= 0.95

    return current
```

```python
values = [1, 4, 7, 5, 3, 6, 9, 8]

for i in range(5):
    result = anneal(values, 0)
    print("Іске қосу", i + 1, ":", values[result])



    #Алгоритмді 5 рет іске қосқанда соңғы мәндер **9, 7, 9, 9, 9** болды. Яғни нәтижелердің барлығы бірдей емес. Себебі Simulated Annealing-де көрші индекс кездейсоқ таңдалады, сондықтан әр іске қосқанда алгоритмнің жүретін жолы өзгеруі мүмкін.
```

```python
# 6 тапсырма

#Hill Climbing** бір бастапқы нүктеден басталады. Ол тек мәні жоғары жаққа өтеді, сондықтан нашар қадамды қабылдамайды. Артықшылығы — қарапайым және тез жұмыс істейді, бірақ local maximum-да тұрып қалуы мүмкін.

#Random Restart** алгоритмді бірнеше әртүрлі бастапқы нүктеден іске қосады. Ол да нашар қадамға өтпейді, бірақ бірнеше жерден бастап көру арқылы жақсы нәтижеге жету мүмкіндігін арттырады. Кемшілігі — алгоритмді бірнеше рет орындау керек.

#Simulated Annealing** көрші күйді кездейсоқ таңдайды және кейде нашар қадамды да қабылдай алады. Сондықтан local maximum-нан шығып кетуге мүмкіндік бар. Бірақ кездейсоқтық болғандықтан, әр іске қосқанда нәтиже әртүрлі болуы мүмкін.
```
